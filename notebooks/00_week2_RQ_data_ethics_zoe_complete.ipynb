## Research Questions and Answers

### 1. Where did the data come from?

The original source was Daqing Chen, from the School of Engineering at London South Bank University. The dataset was donated to UC Irvine Machine Learning Repository on 11/05/2015 (Reasonable interpretation)[1]. The collection method was transactional and observational, covering all transactions between 1 December 2010 and 9 December 2011 for a UK-based and registered non-store online retailer. The company mainly sells unique all-occasion gifts. Many customers of the company are wholesalers. The owner is UC Irvine Machine Learning Repository.

### 2. Why was it originally collected?
The original purpose of this data was to observe the transactions occurring for a UK-based and registered non-store online retailer over a one-year period (reasonable interpretation). The dataset was then used in a study exploring data mining for the online retail industry [2]. 


The dataset was donated after the study was published to the UCI Machine Learning Repository. This is a collection of databases, domain theories, and data generators that are used by the machine learning community for the empirical analysis of machine learning algorithms [3]. 


It is ethically justified because, now that the data is no longer being used for research purposes, it can be used to assist in the development of machine learning. All identifiable variables have been removed; therefore, the ethical commitments have been adhered to. 


### 3. Who is represented?

The dataset represents the transactions from a UK-based online retail company between 1 December 2010 and 9 December 2011. It includes customers from the UK and other countries with many being wholesalers. The data represents individual sales transactions, purchased products, quantities, price, invoice dates and customer information. 


The sampling frame is the company’s recorded online transactions during this period rather than a sample of the general population [1].


### 4. Who might be missing?

 The dataset is missing customers and transactions that were not recorded by the company. An example is people who did not purchase from the retailer during the relevant period. It also may not represent customers of other retailers .


### 5. What personal or sensitive information does it contain, explicitly or implicitly?

CustomerID is a 5 digit unique identifier assigned to each customer in the dataset. The dataset has it in pseudonyms but it acts as a link across all transaction histories for that individual / business . The financial capacity shows a combined transaction totals across the timestamps and it shows total spending, financial balance and cash flow patterns. 


The country provides geographic information which can act as an identifier when paired with unique buying behaviours and timestamps. A few implicit identifiers show behavioural fingerprinting within the transactional logs such as InvoiceDate, StockCode and Quantity. Each of these show the purchasing habits in specific minutes. Specific product purchases such as Description can serve as implicit proxies for personal preferences [1].


The specific combinations of CustomerID, Country, InvoiceDate and Description act as a unique identifier. If these were cross referenced with external data such as public reviews this can identify individuals or small business owners.


### 6. What could your analysis reveal that was not obvious in the original data?

Customer churn can be revealed as calculating the time intervals between InvoiceDate records can predict when a customer is about to stop buying long before they officially churn [4].


It also reveals hidden product and market dynamics. From analysing response rates to pricing changes (UnitPrice) and then to InvoiceDate it reveals the exact purchasing cycles and discount dependences.


The commercial and identification inferences are also revealed when looking through the data. Analysing buying frequency and Quantity can show inventory turnover rates, business size, resale margins and operational health. Also, collecting items purchased at unique times in specific countries can lead to identification of corporate clients or individual buyers. This can uncover private trade secrets or commercial relationships.


### 7. Who could benefit from your findings?

The primary beneficiaries are E-commerce retailers, marketing teams and UX/ UI teams.  
 
Retailers can use these models to help predict people's purchasing power or intent; in addition, they can get a better overview of how to predict consumers' buying habits and thus improve revenue through more personalised targeting [6] 
 
Similarly, marketing teams can use this data to tailor information by creating ‘segments’ that divide customers into distinct groups to focus marketing efforts on more favourable segments [6]. Furthermore, marketing teams use this data to analyse the best times of day to send promotions and communications using the invoice time feature [7].  
 
UX and UI teams can use behavioural predictors such as previous order cancellations to better understand how the customer navigates their retail website. Descriptions of previously bought items can be used to adjust algorithms to push more personalised items [8]. 


### 8. Who could be harmed or misrepresented?

Consumers could be harmed by revealing sensitive health and lifestyle information through their purchasing decisions, without knowing their knowledge or consent. This dataset tracks descriptions of each consumer's purchases, which increases the risk that models are being used to infer personal attributes such as gender, race and age, with inferred demographics being used to target ads and marketing.  [9].
